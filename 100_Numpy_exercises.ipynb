#### 29. How to round away from zero a float array ? (★☆☆)

````python
import numpy as np

# Create a random float array (including negative numbers to test properly)
z = np.random.uniform(-5, 5, 5)
print("Original:", z)

# Round away from zero
result = np.copysign(np.ceil(np.abs(z)), z)
print("Rounded :", result)

# Create a random float array (including negative numbers to test properly)
z = np.random.uniform(-5, 5, 5)
print("Original:", z)

# If greater than 0, round up (ceil); if less than 0, round down (floor)
result = np.where(z > 0, np.ceil(z), np.floor(z))
print("Rounded :", result)
````
